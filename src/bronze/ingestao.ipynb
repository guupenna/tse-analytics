```python
import os

fonte = dbutils.widgets.get("fonte")

volume_path = "/Volumes/tse/raw/arquivos"
folder_path = f"{volume_path}/{fonte}"
files_path = f"{folder_path}/*BRASIL.csv"

df = spark.read.csv(files_path,
                    header=True,
                    sep=";",
                    encoding="latin1",
                    inferSchema=True,
                    multiLine=True)
```

```python
(df.write
    .mode("overwrite")
    .format("delta")
    .partitionBy("ANO_ELEICAO")
    .option("overwriteSchema", "true")
    .saveAsTable(f"tse.bronze.{fonte}"))
```
